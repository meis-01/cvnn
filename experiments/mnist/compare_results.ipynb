# FFT MNIST 8-vs-5: R2 vs Wirtinger

This notebook loads `metrics.json` from the experiment run directory and plots loss/accuracy for the split `R2` and Wirtinger backprop variants.

```python
from pathlib import Path
import json

import matplotlib.pyplot as plt

RUN_DIR = Path('runs/fft_8_vs_5_compare')
METRICS_PATH = RUN_DIR / 'metrics.json'

if not METRICS_PATH.exists():
    raise FileNotFoundError(
        f'{METRICS_PATH} was not found. Run training first, for example: '
        'python -m experiments.mnist.train_fft_mnist --download --epochs 10'
    )

metrics = json.loads(METRICS_PATH.read_text())
metrics
```

## Final/Best Metrics

The current training script stores the best validation epoch metrics plus final test metrics for each method.

```python
methods = ['r2', 'wirtinger']
for method in methods:
    print(method)
    for key, value in metrics[method].items():
        if isinstance(value, float):
            print(f'  {key}: {value:.4f}')
        else:
            print(f'  {key}: {value}')
```

```python
loss_keys = ['train_loss', 'val_loss', 'test_loss']
accuracy_keys = ['train_accuracy', 'val_accuracy', 'test_accuracy']

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

x = range(len(loss_keys))
for method in methods:
    axes[0].plot(x, [metrics[method][key] for key in loss_keys], marker='o', label=method)
axes[0].set_xticks(list(x), loss_keys)
axes[0].set_title('Loss')
axes[0].set_ylabel('Cross entropy')
axes[0].grid(True, alpha=0.3)
axes[0].legend()

x = range(len(accuracy_keys))
for method in methods:
    axes[1].plot(x, [metrics[method][key] for key in accuracy_keys], marker='o', label=method)
axes[1].set_xticks(list(x), accuracy_keys)
axes[1].set_title('Accuracy')
axes[1].set_ylabel('Accuracy')
axes[1].set_ylim(0.0, 1.0)
axes[1].grid(True, alpha=0.3)
axes[1].legend()

fig.tight_layout()
plt.show()
```

## Test Accuracy Comparison

```python
fig, ax = plt.subplots(figsize=(5, 4))
values = [metrics[method]['test_accuracy'] for method in methods]
ax.bar(methods, values, color=['#4C78A8', '#F58518'])
ax.set_ylim(0.0, 1.0)
ax.set_ylabel('Test accuracy')
ax.set_title('FFT MNIST 8 vs 5')
for index, value in enumerate(values):
    ax.text(index, value + 0.01, f'{value:.3f}', ha='center')
ax.grid(True, axis='y', alpha=0.3)
plt.show()
```
